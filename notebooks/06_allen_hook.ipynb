```python
import nflreadpy as nfl
pbp = nfl.load_pbp(2026).to_pandas()

g = pbp[(pbp.home_team.isin(["BUF","MIN"])) & (pbp.away_team.isin(["BUF","MIN"]))]
a = g[(g.posteam=="BUF") & (g.passer_player_name=="J.Allen") & (g.pass_attempt==1)]

print(g[["game_id","week","home_team","away_team"]].drop_duplicates())
print(a.groupby("drive").agg(
    att=("pass_attempt","sum"), comp=("complete_pass","sum"),
    yds=("passing_yards","sum"), td=("pass_touchdown","sum"),
    int_=("interception","sum"), qtr=("qtr","first"),
    secs_left=("game_seconds_remaining","first"),
))
```

```python
a1 = a[a.drive <= 11][["drive","qtr","down","ydstogo","desc","complete_pass","passing_yards"]]
print(a1.to_string())
```
